Import libraries

```python
import PIL.Image as image
import numpy as np
import matplotlib.pyplot as plt
```

Load grayscale image

```python
img = image.open("pcb.tif").convert("L")
img = np.array(img)

H, W = img.shape

plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.show()
```

3×3 Averaging filter

```python
filter = np.array([[1/9, 1/9, 1/9],
                   [1/9, 1/9, 1/9],
                   [1/9, 1/9, 1/9]])
```

zero padding:<br>np.pad(img, pad_width=1, mode='constant', constant_values=0)<br><br>
replicate padding:<br>np.pad(img, pad_width=1, mode='edge')<br><br>
wrap-around padding:<br>np.pad(img, pad_width=1, mode='wrap')<br>

```python
padded_img = np.pad(img, pad_width=1, mode='edge')
```

Apply the 3x3 averaging filter

```python

output1 = np.zeros_like(img)

for i in range(H):
    for j in range(W):
        region = padded_img[i:i+3, j:j+3]
        output1[i, j] = np.sum(region * filter)
```

Apply 3×3 Median filter

```python
output2 = np.zeros_like(img)

for i in range(H):
    for j in range(W):
        region = padded_img[i:i+3, j:j+3]
        output2[i, j] = np.median(region)
```

Showing all images and result comparison

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(output1, cmap='gray')
plt.title("Averaging Filter")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(output2, cmap='gray')
plt.title("Median Filter")
plt.axis("off")

plt.show()
```
